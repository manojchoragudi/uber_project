```python
import pandas as pd

files = [
  {
    "file": "map_cities"
  },
  {
    "file": "map_cancellation_reasons"
  },
  {
    "file": "map_payment_methods"
  },
  {
    "file": "map_ride_statuses"
  },
  {
    "file": "map_vehicle_makes"
  },
  {
    "file": "map_vehicle_types"
  }
]

for file in files:

    url = f"https://uberadls.blob.core.windows.net/raw/git_ingest/{file['file']}.json?sp=r&st=2026-09-01T05:49:33Z&se=2026-09-03T14:04:33Z&spr=https&sv=2026-02-06&sr=c&sig=QNNhDmAtAD%2BbhnsTvWZGCuh75C2RS695CZGeOMiXKDY%3D"

    df = pd.read_json(url)
    df_spark = spark.createDataFrame(df)

    #write data to bronze layer
    df_spark.write.format("delta")\
        .mode("overwrite")\
          .option("overwriteSchema", "true")\
        .saveAsTable(f"uber.bronze.{file['file']}")
```
